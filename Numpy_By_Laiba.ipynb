```python
matrix=np.array([[3,2,1],[9,8,7]])
print(matrix)
```

```python
sum=np.sum(matrix,axis=0)  #column
print(sum)
```

```python
sum=np.sum(matrix,axis=1)  #row
print(sum)
```

```python
mean=np.mean(matrix,axis=1)
print(mean)
```

```python
mean=np.mean(matrix,axis=0)
print(mean)
```

```python
a=np.array([5,13,9])
mean=np.mean(a)
print(mean)
```

```python
a=np.array([5,13,9])
std_dev=np.std(a)
print(std_dev)
```

```python
matrix=np.array([[3,2,1],[9,8,7]])
tranpose=matrix.T
print(tranpose)
```

```python
a=np.array([3,4,5,6,7,8,9])
print(a)
slight_array=a[1:5]
print(slight_array)
```

```python
# Boolean indexing
bool_index = a > 3
print(bool_index)
filter=a[bool_index]
print("ans =",filter)
```

```python
# Concatenate arrays
a=np.array([3,4,5,6,7,8,9])
b=np.array([12,55,200,77])
concat=np.concatenate((a,b))
print(concat)
```

```python
# Stack arrays horizontally
hstack = np.hstack((a, b))
print("Horizontally stacked array:", hstack)
```

```python
# Create identity matrix
identity_matrix =np.eye(4)
print("Identity matrix:\n", identity_matrix)
```

```python
# Random numbers
random_array = np.random.random((3))  # random values in a given shape
print("Random array:\n", random_array)
```
